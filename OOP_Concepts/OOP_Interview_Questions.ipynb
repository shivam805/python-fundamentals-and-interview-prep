## 1. What is `__new__` vs `__init__`?

**`__new__`**: Creates the object (memory allocation)  
**`__init__`**: Initializes the object (setup/configuration)

**Order**: `__new__` → `__init__`

````python
class MyClass:
    def __new__(cls):
        print("Creating object")
        return super().__new__(cls)
    def __init__(self):
        print("Initializing object")

obj = MyClass()  # Output: Creating object, then Initializing object
````

## 2. What is Encapsulation?

**Definition**: Bundling data (attributes) and methods together, hiding internal details

**Access Modifiers**:
- `public`: accessible everywhere
- `_protected`: convention, accessible in subclasses
- `__private`: name-mangled, access limited

````python
class Account:
    def __init__(self, balance):
        self.__balance = balance  # Private
    def deposit(self, amount):
        self.__balance += amount

acc = Account(1000)
acc.deposit(500)
# print(acc.__balance)  # ERROR! Cannot access directly
````

## 3. What is Polymorphism?

**Definition**: "Many forms" - same method name, different behavior based on object

**Types**: Method overloading, method overriding, duck typing

````python
class Animal:
    def speak(self):
        pass

class Dog(Animal):
    def speak(self):
        return "Woof"

class Cat(Animal):
    def speak(self):
        return "Meow"

for animal in [Dog(), Cat()]:
    print(animal.speak())  # Different output, same method name!
````

## 4. Static Methods vs Class Methods

**@staticmethod**: No access to instance or class, just utility function  
**@classmethod**: Access to class (first param is `cls`), factory methods

````python
class MyClass:
    count = 0
    def __init__(self):
        MyClass.count += 1
    
    @staticmethod
    def static_method():
        return "No access to self or cls"
    
    @classmethod
    def class_method(cls):
        return f"Count: {cls.count}"

obj = MyClass()
print(MyClass.static_method())  # Works
print(MyClass.class_method())   # Count: 1
````

## 5. Method Overloading vs Method Overriding

**Overriding**: Child class replaces parent method  
**Overloading**: Python doesn't support true overloading (use `*args`, `**kwargs`)

````python
# Overriding
class Parent:
    def greet(self):
        return "Parent"

class Child(Parent):
    def greet(self):  # Override
        return "Child"

print(Child().greet())  # Child

# Overloading (simulated with *args)
class Math:
    def add(self, *args):
        return sum(args)

m = Math()
print(m.add(1, 2))        # 3
print(m.add(1, 2, 3, 4))  # 10
````

## 6. Inheritance and MRO (Method Resolution Order)

**MRO**: Order Python searches for methods in parent classes  
**Command**: Use `ClassName.__mro__` or `help(ClassName)`

````python
class A:
    def method(self):
        return "A"

class B(A):
    def method(self):
        return "B"

class C(A):
    def method(self):
        return "C"

class D(B, C):
    pass

print(D().method())  # B (left-to-right)
print(D.__mro__)     # Shows: D, B, C, A, object
````

## 7. Abstract Base Classes (ABC)

**Purpose**: Define interface without implementation, force subclasses to implement methods

````python
from abc import ABC, abstractmethod

class Animal(ABC):
    @abstractmethod
    def sound(self):
        pass

class Dog(Animal):
    def sound(self):
        return "Woof"

dog = Dog()
print(dog.sound())  # Woof
# Animal() would raise TypeError - cannot instantiate abstract class
````

## 8. Property Decorators vs Getters/Setters

**@property**: Pythonic way (looks like attribute access)  
**Getter/Setter**: Traditional approach (verbose)

````python
class Circle:
    def __init__(self, radius):
        self._radius = radius
    
    @property
    def area(self):
        return 3.14 * self._radius ** 2
    
    @area.setter
    def area(self, value):
        self._radius = (value / 3.14) ** 0.5

c = Circle(5)
print(c.area)     # Looks like attribute access!
c.area = 100      # Setter called
````

## 9. Multiple Inheritance & Diamond Problem

**Diamond Problem**: Ambiguity when class inherits from two classes that share a parent  
**Solution**: Use `super()` and C3 Linearization (MRO)

````python
class A:
    def method(self):
        print("A")

class B(A):
    def method(self):
        super().method()
        print("B")

class C(A):
    def method(self):
        super().method()
        print("C")

class D(B, C):
    pass

D().method()  # A, C, B (follows MRO)
````

## 10. Composition vs Inheritance

**Inheritance**: IS-A relationship (Dog IS-A Animal)  
**Composition**: HAS-A relationship (Car HAS-A Engine)

````python
# Inheritance (IS-A)
class Animal:
    def speak(self):
        pass

class Dog(Animal):
    def speak(self):
        return "Woof"

# Composition (HAS-A) - Better approach!
class Engine:
    def start(self):
        return "Engine started"

class Car:
    def __init__(self):
        self.engine = Engine()  # HAS-A engine
    
    def start(self):
        return self.engine.start()

car = Car()
print(car.start())  # Engine started
````

## 11. What is `super()`?

**Purpose**: Call parent class method without explicitly naming it  
**Benefit**: Works with multiple inheritance (uses MRO)

````python
class Parent:
    def __init__(self, name):
        self.name = name

class Child(Parent):
    def __init__(self, name, age):
        super().__init__(name)  # Call parent's __init__
        self.age = age

child = Child("Alice", 10)
print(f"{child.name}, {child.age}")  # Alice, 10
````

## 12. What is `isinstance()` and `issubclass()`?

**isinstance(obj, Class)**: Check if object is instance of class  
**issubclass(Class1, Class2)**: Check if Class1 inherits from Class2

````python
class Animal:
    pass

class Dog(Animal):
    pass

dog = Dog()
print(isinstance(dog, Dog))      # True
print(isinstance(dog, Animal))   # True
print(issubclass(Dog, Animal))   # True
print(isinstance("hello", Dog))  # False
````

# 13. PUBLIC, PRIVATE, and PROTECTED Attributes

**Definition**: Control who can access class attributes and methods

## Access Levels:

| Modifier | Syntax | Access From | Use Case |
|----------|--------|-------------|----------|
| **Public** | `name` | Anywhere | Intended for external use |
| **Protected** | `_name` | Class & Subclasses | Internal implementation, hint to subclass |
| **Private** | `__name` | Only inside class | Hide implementation details |

**Python Note**: No true access control! These are conventions and name-mangling.

````python
# Example 1: PUBLIC - Accessible Everywhere
class Person:
    def __init__(self, name):
        self.name = name  # PUBLIC

person = Person("Alice")
print(person.name)  # Alice - Anyone can access!
person.name = "Bob"  # Anyone can modify!
````

````python
# Example 2: PROTECTED - Hint: "Use in subclasses"
class Vehicle:
    def __init__(self, speed):
        self._speed = speed  # PROTECTED
    
    def _accelerate(self):  # PROTECTED method
        self._speed += 10

class Car(Vehicle):
    def boost(self):
        self._accelerate()  # Subclass CAN access

car = Car(50)
car.boost()
print(car._speed)  # 60 - Works, but using convention "_" means "be careful"
````

````python
# Example 3: PRIVATE - Hidden from Outside
class BankAccount:
    def __init__(self, balance):
        self.__balance = balance  # PRIVATE (name-mangled)
    
    def deposit(self, amount):
        self.__balance += amount
    
    def get_balance(self):
        return self.__balance

account = BankAccount(1000)
print(account.get_balance())  # 1000
account.deposit(500)
# print(account.__balance)  # ERROR! Cannot access directly!
print(account._BankAccount__balance)  # Name-mangled! Don't use this!
````

````python
# Example 4: Comparison - All Three Together
class Employee:
    company = "TechCorp"  # PUBLIC class variable
    
    def __init__(self, name, salary):
        self.name = name              # PUBLIC
        self._department = "IT"       # PROTECTED
        self.__salary = salary        # PRIVATE
    
    def public_method(self):
        return f"{self.name} works here"
    
    def _protected_method(self):
        return f"Internal: {self._department}"
    
    def __private_method(self):
        return f"Secret: {self.__salary}"
    
    def get_info(self):  # PUBLIC method that uses private
        return self.__private_method()

emp = Employee("Alice", 50000)
print(emp.name)                # ✓ Works - PUBLIC
print(emp.public_method())     # ✓ Works - PUBLIC method
print(emp._protected_method()) # ⚠ Works but should only use in subclass
# print(emp.__private_method()) # ✗ ERROR - PRIVATE
print(emp.get_info())          # ✓ Works - Uses private internally
````

---

## Key Points to Remember:

### PUBLIC (`name`)
- No prefix
- Accessible everywhere: outside class, subclasses, instances
- Use for: Interface you want to expose

### PROTECTED (`_name`)
- Single underscore prefix
- Convention: "This is for internal use, mainly for subclasses"
- Python doesn't enforce! Anyone can still access
- Use for: Methods/attributes meant for subclass overriding

### PRIVATE (`__name`)
- Double underscore prefix
- **Name mangled** to `_ClassName__name`
- Hardest to access from outside
- Use for: Implementation details you want to hide

### Important:
> ⚠️ **Python doesn't truly enforce privacy!** These are conventions and name-mangling tricks. A determined programmer can still access "private" members. Use encapsulation + methods for true data hiding.

````python
# Example 5: Name Mangling Explained
class Secret:
    def __init__(self):
        self.__secret = "Hidden!"

obj = Secret()
# obj.__secret  # ERROR - attribute doesn't exist

# But Python renames it internally:
print(obj._Secret__secret)  # Works! But DON'T use this in real code

# Show all attributes
print(dir(obj))  # You'll see _Secret__secret in the list
````

# 14. METACLASSES

**Definition**: A metaclass is a "class of a class" - it defines how a class behaves. Just as an object is an instance of a class, a class is an instance of a metaclass.

**Key Points**:
- All classes are instances of `type` (the default metaclass)
- Metaclasses execute when the CLASS is defined, not when instances are created
- Use `__new__` to customize class creation
- Real-world: Django models, SQLAlchemy ORM, API frameworks

**Key Question**: When does it run?
- **Metaclass executes**: When class is DEFINED
- **Regular __init__ executes**: When instance is created

````python
# Example 1: Understanding type() - The Default Metaclass
class Animal:
    def speak(self):
        return "Sound"

print(f"Type of Animal: {type(Animal)}")  # <class 'type'>
print(f"Type of instance: {type(Animal())}")  # <class '__main__.Animal'>

# Create class dynamically with type()
Dog = type('Dog', (Animal,), {'speak': lambda self: "Woof!"})
print(f"Dog says: {Dog().speak()}")  # Woof!
````

````python
# Example 2: Custom Metaclass - Auto-Logging
class LogMeta(type):
    def __new__(mcs, name, bases, dct):
        for key, value in dct.items():
            if callable(value) and not key.startswith('_'):
                dct[key] = lambda self, v=value: print(f"Calling {key}") or v(self)
        return super().__new__(mcs, name, bases, dct)

class Calculator(metaclass=LogMeta):
    def add(self, a, b):
        return a + b

calc = Calculator()
calc.add(2, 3)  # Prints: "Calling add"
````

````python
# Example 3: Singleton Pattern via Metaclass
class SingletonMeta(type):
    _instances = {}
    def __call__(cls, *args, **kwargs):
        if cls not in cls._instances:
            cls._instances[cls] = super().__call__(*args, **kwargs)
        return cls._instances[cls]

class Database(metaclass=SingletonMeta):
    def __init__(self):
        self.connection = "Connected"

db1 = Database()
db2 = Database()
print(f"Same instance: {db1 is db2}")  # True
````

### Key Interview Points:

**Execution Timing:**
- Metaclass `__new__` executes when CLASS is defined
- Regular `__init__` executes when INSTANCE is created
- Metaclass `__call__` executes when instance is created

**Real-World Uses:**
- Django models (defines database schema)
- SQLAlchemy ORM (maps Python classes to DB tables)
- APIs with validation at class definition time
- Singleton pattern for resource management
- Abstract base class enforcement

**Interview Answer Template:**
"A metaclass is a 'class of a class' - it defines how a class behaves. The default metaclass is `type`. We create custom metaclasses by inheriting from `type` and overriding `__new__` to customize class creation. Common use case: Django uses metaclasses to convert model classes into ORM structures at definition time."
